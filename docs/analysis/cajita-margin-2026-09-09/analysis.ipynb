## Results

```python
for r in m['scenarios']:
    print(f"{r['scenario']} | {r['product']} | cost ${r['total_variable_cost']:.2f} | contribution ${r['contribution']:.2f} | margin {r['contribution_margin']:.1%}")
for price, minutes, packaging in [(18,12,m['base_pack']), (20,14,m['base_pack']+.35)]:
    r=m['calculate'](price,minutes=minutes,packaging=packaging)
    print(f"Alternative ${price}: {r['contribution_margin']:.1%} before fixed overhead")
```
